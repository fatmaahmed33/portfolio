```python
import numpy as np
import pandas as pd

df = pd.read_csv("/content/Dataset.csv")

df.head()
```

```python
df.dtypes
```

```python
col_summary = pd.DataFrame(
    {
        "Data Type": df.dtypes,
        "Missing Values": df.isnull().sum(),
        "Missing Ratio (%)": (df.isnull().sum() / len(df)) * 100,
        "Unique Values": df.nunique(),
    }
)

col_summary
```

```python
df.describe().T
```

```python
import numpy as np

df.loc[df["Quantity"] < 0, "Quantity"] = np.nan
df.loc[df["Sales"] < 0, "Sales"] = np.nan
df.loc[df["OrderValue"] < 0, "OrderValue"] = np.nan

df.isnull().sum()
```

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Sales"] = df["Sales"].fillna(df["Sales"].median())
df["OrderValue"] = df["OrderValue"].fillna(df["OrderValue"].median())

df.isnull().sum()
```

```python
df.describe(include=["object"]).T
```

```python
df['OrderDate'] =pd.to_datetime(df['OrderDate'])
df['SignupDate']=pd.to_datetime(df['SignupDate'])
```

```python
df.describe(include=["object"]).T
```

```python


df["Category"].value_counts()
```

```python
df.drop_duplicates(inplace=True)

df.columns = df.columns.str.strip()

df.isnull().sum()
```

```python
df.groupby("Category")["Sales"].mean().sort_values(ascending=False)
```

```python
df.groupby("Status")["Sales"].mean().sort_values(ascending=False)
```

```python
df.groupby("CustomerSegment")["Sales"].mean().sort_values(ascending=False)
```

```python
df.groupby("PaymentMethod")["Sales"].mean().sort_values(ascending=False)
```

```python
num_df = df[["Quantity", "Discount", "Age", "UnitPrice", "Sales"]]

q1 = num_df.quantile(0.25)
q3 = num_df.quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers_count = ((num_df < lower_bound) | (num_df > upper_bound)).sum()
outliers_df = pd.DataFrame(
    {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outliers Count": outliers_count,
    }
)

outliers_df
```

```python

remaining_cats = df.select_dtypes(include="object").columns.tolist()

df = pd.get_dummies(df, columns=remaining_cats, drop_first=False)

df.head()
```

```python
import matplotlib.pyplot as plt

df_plot = pd.read_csv("Dataset.csv")
df_plot["OrderDate"] = pd.to_datetime(df_plot["OrderDate"])

monthly_sales = df_plot.groupby(df_plot["OrderDate"].dt.to_period("M"))[
    "Sales"
].sum()

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o",
    linestyle="--",
    linewidth=2,
    color="red",
)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.show()
```

```python
category_counts = df_plot["Category"].value_counts()

plt.bar(category_counts.index, category_counts.values, color="teal")
plt.title("Number of Orders by Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
status_counts = df_plot["Status"].value_counts()

plt.pie(
    status_counts.values,
    labels=status_counts.index,
    autopct="%1.1f%%",
    colors=["green", "orange", "red"],
)
plt.title("Order Status Distribution")
plt.show()
```

```python
plt.hist(df_plot["Age"], bins=15, color="skyblue", edgecolor="black")
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.boxplot(df_plot["Sales"], vert=False)
plt.title("Sales Outliers Detection")
plt.xlabel("Sales")
plt.show()
```

```python
plt.scatter(
    df_plot["UnitPrice"],
    df_plot["Sales"],
    marker="o",
    color="purple",
    alpha=0.1,
)
plt.title("Relationship between Unit Price and Sales")
plt.xlabel("Unit Price")
plt.ylabel("Sales")
plt.show()
```
